```python
import pandas as pd

# Cargar 2024
df_2024 = pd.read_excel('/content/Bogota_ocu 2024 N VIVO.xlsx')
print(f"Registros 2024: {len(df_2024):,}")

# Cargar 2023
df_2023 = pd.read_excel('/content/Bogota_ocu 2023N VIVO .xlsx')
print(f"Registros 2023: {len(df_2023):,}")

# Ver columnas
print(f"\nColumnas disponibles:")
for i, col in enumerate(df_2024.columns):
    print(f"{i:2d}. {col}")
```

```python
# ANÁLISIS EXPLORATORIO COMPLETO
print("=" * 70)
print("ANÁLISIS EXPLORATORIO - DATOS REALES COMPLETOS")
print("=" * 70)

# 1. Clasificar por edad gestacional
def clasificar_eg(sem):
    if pd.isna(sem):
        return 'Sin dato'
    elif sem < 32:
        return '1. Muy prematuro (<32 sem)'
    elif sem < 37:
        return '2. Prematuro (32-36 sem)'
    else:
        return '3. A término (>=37 sem)'

df_2024['categoria_eg'] = df_2024['T_GES'].apply(clasificar_eg)

# 2. Distribución por categoría
print("\n--- DISTRIBUCIÓN POR EDAD GESTACIONAL (2024) ---")
dist = df_2024['categoria_eg'].value_counts().sort_index()
print(dist)
print(f"\nPorcentajes:")
print((dist / len(df_2024) * 100).round(2))

# 3. Filtrar solo residentes Bogotá (código 11)
df_bogota = df_2024[df_2024['CODPTORE'] == 11]
print(f"\n--- RESIDENTES BOGOTÁ ---")
print(f"Total Bogotá: {len(df_bogota):,} ({len(df_bogota)/len(df_2024)*100:.1f}%)")

# 4. Distribución en residentes Bogotá
print("\n--- EDAD GESTACIONAL EN RESIDENTES BOGOTÁ ---")
dist_bog = df_bogota['categoria_eg'].value_counts().sort_index()
print(dist_bog)
print(f"\nPorcentajes:")
print((dist_bog / len(df_bogota) * 100).round(2))

# 5. Promedios mensuales
print("\n--- PROMEDIOS MENSUALES (Residentes Bogotá) ---")
muy_prem = len(df_bogota[df_bogota['T_GES'] < 32])
prem = len(df_bogota[(df_bogota['T_GES'] >= 32) & (df_bogota['T_GES'] < 37)])
termino = len(df_bogota[df_bogota['T_GES'] >= 37])

print(f"Muy prematuros (<32 sem): {muy_prem:,} anuales → {muy_prem/12:.0f}/mes")
print(f"Prematuros (32-36 sem): {prem:,} anuales → {prem/12:.0f}/mes")
print(f"A término (>=37 sem): {termino:,} anuales → {termino/12:.0f}/mes")

# 6. Por mes
print("\n--- NACIMIENTOS POR MES (Residentes Bogotá) ---")
tabla = pd.crosstab(df_bogota['MES'], df_bogota['categoria_eg'])
print(tabla)
```

```python
# =============================================================================
# ANÁLISIS EXPLORATORIO DE ALTO NIVEL - VALIDACIÓN DE DATOS
# Bases Nacidos Vivos Bogotá 2023-2024
# =============================================================================

import pandas as pd
import numpy as np

print("=" * 80)
print("ANÁLISIS EXPLORATORIO DE ALTO NIVEL - VALIDACIÓN PARA SIMULACIONES")
print("=" * 80)

# =============================================================================
# 1. ESTRUCTURA Y COMPLETITUD DE DATOS
# =============================================================================
print("\n" + "=" * 80)
print("1. ESTRUCTURA Y COMPLETITUD DE DATOS")
print("=" * 80)

print(f"\n--- DIMENSIONES ---")
print(f"Base 2024: {df_2024.shape[0]:,} registros × {df_2024.shape[1]} columnas")
print(f"Base 2023: {df_2023.shape[0]:,} registros × {df_2023.shape[1]} columnas")

# Columnas prioritarias para el análisis
cols_prioritarias = ['T_GES', 'MES', 'ANO', 'CODPTORE', 'CODMUNRE', 'PESO_NAC',
                     'SEXO', 'EDAD_MADRE', 'TIPO_PARTO', 'MUL_PARTO']

print(f"\n--- VALORES FALTANTES EN COLUMNAS PRIORITARIAS (2024) ---")
for col in cols_prioritarias:
    if col in df_2024.columns:
        faltantes = df_2024[col].isna().sum()
        pct = (faltantes / len(df_2024)) * 100
        estado = "OK" if pct < 1 else "REVISAR" if pct < 5 else "ALERTA"
        print(f"{col:15} → Faltantes: {faltantes:,} ({pct:.2f}%) [{estado}]")

print(f"\n--- VALORES FALTANTES EN COLUMNAS PRIORITARIAS (2023) ---")
for col in cols_prioritarias:
    if col in df_2023.columns:
        faltantes = df_2023[col].isna().sum()
        pct = (faltantes / len(df_2023)) * 100
        estado = "OK" if pct < 1 else "REVISAR" if pct < 5 else "ALERTA"
        print(f"{col:15} → Faltantes: {faltantes:,} ({pct:.2f}%) [{estado}]")

# =============================================================================
# 2. ANÁLISIS DE EDAD GESTACIONAL (T_GES)
# =============================================================================
print("\n" + "=" * 80)
print("2. ANÁLISIS DE EDAD GESTACIONAL (T_GES)")
print("=" * 80)

print(f"\n--- ESTADÍSTICAS DESCRIPTIVAS ---")
print(f"{'Estadística':<20} {'2024':>12} {'2023':>12}")
print("-" * 46)
print(f"{'Media':<20} {df_2024['T_GES'].mean():>12.2f} {df_2023['T_GES'].mean():>12.2f}")
print(f"{'Mediana':<20} {df_2024['T_GES'].median():>12.1f} {df_2023['T_GES'].median():>12.1f}")
print(f"{'Desv. Estándar':<20} {df_2024['T_GES'].std():>12.2f} {df_2023['T_GES'].std():>12.2f}")
print(f"{'Mínimo':<20} {df_2024['T_GES'].min():>12.0f} {df_2023['T_GES'].min():>12.0f}")
print(f"{'Máximo':<20} {df_2024['T_GES'].max():>12.0f} {df_2023['T_GES'].max():>12.0f}")

print(f"\n--- VALORES ATÍPICOS ---")
# Valores fuera de rango esperado (20-44 semanas)
atipicos_2024 = len(df_2024[(df_2024['T_GES'] < 20) | (df_2024['T_GES'] > 44)])
atipicos_2023 = len(df_2023[(df_2023['T_GES'] < 20) | (df_2023['T_GES'] > 44)])
print(f"Fuera de rango (20-44 sem) 2024: {atipicos_2024} ({atipicos_2024/len(df_2024)*100:.3f}%)")
print(f"Fuera de rango (20-44 sem) 2023: {atipicos_2023} ({atipicos_2023/len(df_2023)*100:.3f}%)")

print(f"\n--- DISTRIBUCIÓN POR CATEGORÍA (Comparativo) ---")
def clasificar_eg(sem):
    if pd.isna(sem): return '0. Sin dato'
    elif sem < 28: return '1. Extremadamente prematuro (<28)'
    elif sem < 32: return '2. Muy prematuro (28-31)'
    elif sem < 34: return '3. Prematuro moderado (32-33)'
    elif sem < 37: return '4. Prematuro tardío (34-36)'
    elif sem < 39: return '5. Término temprano (37-38)'
    elif sem <= 41: return '6. Término completo (39-41)'
    else: return '7. Postérmino (>41)'

df_2024['cat_eg_detalle'] = df_2024['T_GES'].apply(clasificar_eg)
df_2023['cat_eg_detalle'] = df_2023['T_GES'].apply(clasificar_eg)

dist_2024 = df_2024['cat_eg_detalle'].value_counts().sort_index()
dist_2023 = df_2023['cat_eg_detalle'].value_counts().sort_index()

print(f"\n{'Categoría':<35} {'2024':>8} {'%':>7} {'2023':>8} {'%':>7}")
print("-" * 70)
for cat in dist_2024.index:
    n24 = dist_2024.get(cat, 0)
    n23 = dist_2023.get(cat, 0)
    p24 = n24/len(df_2024)*100
    p23 = n23/len(df_2023)*100
    print(f"{cat:<35} {n24:>8,} {p24:>6.2f}% {n23:>8,} {p23:>6.2f}%")

# =============================================================================
# 3. ANÁLISIS DE SERIE TEMPORAL (ESTABILIDAD MENSUAL)
# =============================================================================
print("\n" + "=" * 80)
print("3. ANÁLISIS DE SERIE TEMPORAL - ESTABILIDAD MENSUAL")
print("=" * 80)

# Nacimientos por mes
nac_mes_2024 = df_2024.groupby('MES').size()
nac_mes_2023 = df_2023.groupby('MES').size()

print(f"\n--- NACIMIENTOS POR MES ---")
print(f"{'Mes':<6} {'2024':>10} {'2023':>10} {'Diferencia':>12} {'Var %':>10}")
print("-" * 52)
for mes in range(1, 13):
    n24 = nac_mes_2024.get(mes, 0)
    n23 = nac_mes_2023.get(mes, 0)
    diff = n24 - n23
    var_pct = ((n24 - n23) / n23 * 100) if n23 > 0 else 0
    print(f"{mes:<6} {n24:>10,} {n23:>10,} {diff:>+12,} {var_pct:>+9.1f}%")

print(f"\n{'TOTAL':<6} {nac_mes_2024.sum():>10,} {nac_mes_2023.sum():>10,} {nac_mes_2024.sum()-nac_mes_2023.sum():>+12,}")

print(f"\n--- COEFICIENTE DE VARIACIÓN MENSUAL ---")
cv_2024 = (nac_mes_2024.std() / nac_mes_2024.mean()) * 100
cv_2023 = (nac_mes_2023.std() / nac_mes_2023.mean()) * 100
print(f"CV 2024: {cv_2024:.2f}% {'(Estable)' if cv_2024 < 10 else '(Variable)'}")
print(f"CV 2023: {cv_2023:.2f}% {'(Estable)' if cv_2023 < 10 else '(Variable)'}")

# =============================================================================
# 4. ANÁLISIS DE RESIDENCIA GEOGRÁFICA
# =============================================================================
print("\n" + "=" * 80)
print("4. ANÁLISIS DE RESIDENCIA GEOGRÁFICA")
print("=" * 80)

print(f"\n--- DISTRIBUCIÓN POR DEPARTAMENTO DE RESIDENCIA (2024) ---")
dept_2024 = df_2024['CODPTORE'].value_counts().head(10)
print(f"{'Código Dpto':<15} {'Registros':>12} {'%':>10}")
print("-" * 40)
for cod, n in dept_2024.items():
    nombre = "BOGOTÁ D.C." if cod == 11 else "CUNDINAMARCA" if cod == 25 else f"Otro ({cod})"
    print(f"{nombre:<15} {n:>12,} {n/len(df_2024)*100:>9.2f}%")

# Filtrar solo Bogotá
df_bog_2024 = df_2024[df_2024['CODPTORE'] == 11]
df_bog_2023 = df_2023[df_2023['CODPTORE'] == 11]

print(f"\n--- RESUMEN RESIDENTES BOGOTÁ ---")
print(f"2024: {len(df_bog_2024):,} de {len(df_2024):,} ({len(df_bog_2024)/len(df_2024)*100:.1f}%)")
print(f"2023: {len(df_bog_2023):,} de {len(df_2023):,} ({len(df_bog_2023)/len(df_2023)*100:.1f}%)")

# =============================================================================
# 5. CORRELACIONES Y CONSISTENCIA INTERNA
# =============================================================================
print("\n" + "=" * 80)
print("5. CORRELACIONES Y CONSISTENCIA INTERNA")
print("=" * 80)

print(f"\n--- RELACIÓN EDAD GESTACIONAL vs PESO AL NACER (2024) ---")
# Peso promedio por categoría de EG
peso_por_eg = df_2024.groupby('cat_eg_detalle')['PESO_NAC'].agg(['mean', 'std', 'count'])
print(f"\n{'Categoría EG':<35} {'Peso medio':>12} {'DE':>10} {'N':>8}")
print("-" * 70)
for cat in peso_por_eg.index:
    media = peso_por_eg.loc[cat, 'mean']
    de = peso_por_eg.loc[cat, 'std']
    n = peso_por_eg.loc[cat, 'count']
    print(f"{cat:<35} {media:>12,.0f}g {de:>9,.0f}g {n:>8,}")

# Correlación
corr = df_2024[['T_GES', 'PESO_NAC']].corr().iloc[0,1]
print(f"\nCorrelación T_GES - PESO_NAC: {corr:.3f} {'(Esperada: positiva fuerte)' if corr > 0.5 else ''}")

# =============================================================================
# 6. ANÁLISIS ESPECÍFICO PARA SIMULACIÓN NIRSEVIMAB
# =============================================================================
print("\n" + "=" * 80)
print("6. ANÁLISIS ESPECÍFICO PARA SIMULACIÓN NIRSEVIMAB")
print("=" * 80)

# Solo residentes Bogotá
df_bog_2024['categoria_eg'] = df_bog_2024['T_GES'].apply(lambda x:
    'Muy prematuro (<32)' if x < 32 else
    'Prematuro (32-36)' if x < 37 else
    'A término (>=37)')

print(f"\n--- DISTRIBUCIÓN MENSUAL POR CATEGORÍA - RESIDENTES BOGOTÁ 2024 ---")
tabla_mes_cat = pd.crosstab(df_bog_2024['MES'], df_bog_2024['categoria_eg'])
print(tabla_mes_cat)

print(f"\n--- ESTADÍSTICAS MENSUALES POR CATEGORÍA ---")
for cat in ['Muy prematuro (<32)', 'Prematuro (32-36)', 'A término (>=37)']:
    serie = tabla_mes_cat[cat]
    print(f"\n{cat}:")
    print(f"  Media mensual: {serie.mean():.1f}")
    print(f"  Desv. Est.: {serie.std():.1f}")
    print(f"  Mínimo: {serie.min()} (mes {serie.idxmin()})")
    print(f"  Máximo: {serie.max()} (mes {serie.idxmax()})")
    print(f"  CV: {serie.std()/serie.mean()*100:.1f}%")

# =============================================================================
# 7. CONCLUSIONES PARA SIMULACIÓN
# =============================================================================
print("\n" + "=" * 80)
print("7. CONCLUSIONES PARA SIMULACIÓN")
print("=" * 80)

muy_prem = len(df_bog_2024[df_bog_2024['T_GES'] < 32])
prem = len(df_bog_2024[(df_bog_2024['T_GES'] >= 32) & (df_bog_2024['T_GES'] < 37)])
termino = len(df_bog_2024[df_bog_2024['T_GES'] >= 37])
total = len(df_bog_2024)

print(f"""
POBLACIÓN OBJETIVO (Residentes Bogotá 2024):
  - Total nacimientos: {total:,}
  - Muy prematuros (<32 sem): {muy_prem:,} ({muy_prem/total*100:.2f}%)
  - Prematuros (32-36 sem): {prem:,} ({prem/total*100:.2f}%)
  - A término (>=37 sem): {termino:,} ({termino/total*100:.2f}%)

PROMEDIOS MENSUALES:
  - Muy prematuros: {muy_prem/12:.0f}/mes
  - Prematuros: {prem/12:.0f}/mes
  - A término: {termino/12:.0f}/mes

VALIDACIONES:
  - Datos faltantes: {'MÍNIMOS' if df_2024['T_GES'].isna().sum() < 100 else 'REVISAR'}
  - Consistencia EG-Peso: {'OK' if corr > 0.5 else 'REVISAR'}
  - Estabilidad temporal: {'OK' if cv_2024 < 10 else 'REVISAR'}
  - Comparabilidad 2023-2024: {'OK' if abs(len(df_2024)-len(df_2023))/len(df_2023) < 0.15 else 'DIFERENCIA NOTABLE'}

CONCLUSIÓN: {'DATOS APTOS PARA SIMULACIÓN' if corr > 0.5 and cv_2024 < 15 else 'REVISAR ANTES DE SIMULAR'}
""")

print("=" * 80)
print("ANÁLISIS EXPLORATORIO COMPLETADO")
print("=" * 80)
```

```python
# Filtrar valores atípicos de T_GES
print("--- LIMPIEZA DE DATOS ATÍPICOS ---")
print(f"Registros originales: {len(df_bog_2024):,}")

# Excluir T_GES fuera de rango biológico plausible (22-44 semanas)
df_bog_limpio = df_bog_2024[(df_bog_2024['T_GES'] >= 22) & (df_bog_2024['T_GES'] <= 44)]
print(f"Registros después de filtrar T_GES 22-44: {len(df_bog_limpio):,}")
print(f"Excluidos: {len(df_bog_2024) - len(df_bog_limpio)}")

# Verificar distribución final
print("\n--- DISTRIBUCIÓN FINAL PARA SIMULACIÓN ---")
muy_prem = len(df_bog_limpio[df_bog_limpio['T_GES'] < 32])
prem = len(df_bog_limpio[(df_bog_limpio['T_GES'] >= 32) & (df_bog_limpio['T_GES'] < 37)])
termino = len(df_bog_limpio[df_bog_limpio['T_GES'] >= 37])

print(f"Muy prematuros (<32): {muy_prem:,} → {muy_prem/12:.0f}/mes")
print(f"Prematuros (32-36): {prem:,} → {prem/12:.0f}/mes")
print(f"A término (≥37): {termino:,} → {termino/12:.0f}/mes")
```

```python
# =============================================================================
# CÁLCULO DE 4 ESCENARIOS DE DEMANDA NIRSEVIMAB
# Datos validados - Residentes Bogotá 2024
# =============================================================================

print("=" * 70)
print("ESCENARIOS DE DEMANDA NIRSEVIMAB - DATOS VALIDADOS")
print("=" * 70)

# PARÁMETROS
cobertura_vacunal = 0.80  # Meta 80%
meses_temporada = 4       # Nov 2025 - Feb 2026
dosis_disponibles = 6500

# Promedios mensuales validados
muy_prem_mes = 75
prem_mes = 476
termino_mes = 4111

print(f"""
PARÁMETROS DEL MODELO:
  - Cobertura vacunal materna: {cobertura_vacunal*100:.0f}%
  - Temporada RSV: Nov 2025 - Feb 2026 ({meses_temporada} meses)
  - Dosis disponibles: {dosis_disponibles:,}
""")

# =============================================================================
# ESCENARIO 1: NO INMUNIZADOS (Referencia)
# =============================================================================
print("=" * 70)
print("ESCENARIO 1: NO INMUNIZADOS (Grupo de referencia)")
print("=" * 70)
print("Madre NO vacunada + RN NO recibe nirsevimab")
print("→ Este grupo NO requiere dosis, es la referencia para medir efectividad")

# =============================================================================
# ESCENARIO 2: SOLO VACUNACIÓN MATERNA
# =============================================================================
print("\n" + "=" * 70)
print("ESCENARIO 2: SOLO VACUNACIÓN MATERNA (RSVpreF)")
print("=" * 70)
print("Madre SÍ vacunada + RN a término + SIN condiciones de riesgo")
print("→ Protegidos por anticuerpos transplacentarios, NO requieren nirsevimab")

esc2_mes = termino_mes * cobertura_vacunal
print(f"\nPoblación protegida por vacuna materna: {esc2_mes:.0f}/mes")
print("Dosis nirsevimab requeridas: 0")

# =============================================================================
# ESCENARIO 3: SOLO NIRSEVIMAB (Madres NO vacunadas)
# =============================================================================
print("\n" + "=" * 70)
print("ESCENARIO 3: SOLO NIRSEVIMAB (Hijos de madres NO vacunadas)")
print("=" * 70)

prop_no_vacunada = 1 - cobertura_vacunal

esc3a_mes = muy_prem_mes * prop_no_vacunada
esc3b_mes = prem_mes * prop_no_vacunada
esc3c_mes = termino_mes * prop_no_vacunada

print(f"""
Madre NO vacunada (20%) → RN requiere nirsevimab:

  3A. Muy prematuros (<32 sem):  {muy_prem_mes} × {prop_no_vacunada:.0%} = {esc3a_mes:.0f} dosis/mes
  3B. Prematuros (32-36 sem):    {prem_mes} × {prop_no_vacunada:.0%} = {esc3b_mes:.0f} dosis/mes
  3C. A término (≥37 sem):       {termino_mes} × {prop_no_vacunada:.0%} = {esc3c_mes:.0f} dosis/mes

  Subtotal Escenario 3: {esc3a_mes + esc3b_mes + esc3c_mes:.0f} dosis/mes
""")

# =============================================================================
# ESCENARIO 4: AMBAS INTERVENCIONES (Prematuros de madres vacunadas)
# =============================================================================
print("=" * 70)
print("ESCENARIO 4: AMBAS INTERVENCIONES (Vacuna materna + Nirsevimab)")
print("=" * 70)

esc4a_mes = muy_prem_mes * cobertura_vacunal
esc4b_mes = prem_mes * cobertura_vacunal

print(f"""
Madre SÍ vacunada (80%) + RN prematuro → Requiere nirsevimab adicional:
(Transferencia transplacentaria insuficiente en prematuros)

  4A. Muy prematuros (<32 sem):  {muy_prem_mes} × {cobertura_vacunal:.0%} = {esc4a_mes:.0f} dosis/mes
  4B. Prematuros (32-36 sem):    {prem_mes} × {cobertura_vacunal:.0%} = {esc4b_mes:.0f} dosis/mes

  Subtotal Escenario 4: {esc4a_mes + esc4b_mes:.0f} dosis/mes
""")

# =============================================================================
# RESUMEN CONSOLIDADO
# =============================================================================
print("=" * 70)
print("RESUMEN CONSOLIDADO DE DEMANDA")
print("=" * 70)

# Calcular totales
demanda_prioritaria_mes = esc3a_mes + esc3b_mes + esc4a_mes + esc4b_mes
demanda_adicional_mes = esc3c_mes
demanda_total_mes = demanda_prioritaria_mes + demanda_adicional_mes

demanda_prioritaria_temp = demanda_prioritaria_mes * meses_temporada
demanda_adicional_temp = demanda_adicional_mes * meses_temporada
demanda_total_temp = demanda_total_mes * meses_temporada

print(f"""
DEMANDA MENSUAL:
┌─────────────────────────────────────────────────────────────┐
│ PRIORITARIA (Todos los prematuros)                          │
│   Escenario 3A (Muy prem, madre NO vac):        {esc3a_mes:>6.0f}      │
│   Escenario 3B (Prem, madre NO vac):            {esc3b_mes:>6.0f}      │
│   Escenario 4A (Muy prem, madre SÍ vac):        {esc4a_mes:>6.0f}      │
│   Escenario 4B (Prem, madre SÍ vac):            {esc4b_mes:>6.0f}      │
│   SUBTOTAL PRIORITARIA:                         {demanda_prioritaria_mes:>6.0f}/mes  │
├─────────────────────────────────────────────────────────────┤
│ ADICIONAL (A término de madre no vacunada)                  │
│   Escenario 3C:                                 {demanda_adicional_mes:>6.0f}/mes  │
├─────────────────────────────────────────────────────────────┤
│ TOTAL MENSUAL:                                  {demanda_total_mes:>6.0f}/mes  │
└─────────────────────────────────────────────────────────────┘

DEMANDA TEMPORADA RSV ({meses_temporada} meses: Nov 2025 - Feb 2026):
┌─────────────────────────────────────────────────────────────┐
│ Demanda prioritaria:                          {demanda_prioritaria_temp:>7,.0f} dosis │
│ Demanda adicional:                            {demanda_adicional_temp:>7,.0f} dosis │
│ DEMANDA TOTAL TEMPORADA:                      {demanda_total_temp:>7,.0f} dosis │
└─────────────────────────────────────────────────────────────┘
""")

# =============================================================================
# EVALUACIÓN DE COBERTURA
# =============================================================================
print("=" * 70)
print("EVALUACIÓN DE COBERTURA CON DOSIS DISPONIBLES")
print("=" * 70)

print(f"""
Dosis disponibles (marzo 2026):     {dosis_disponibles:>7,}
Demanda total temporada:            {demanda_total_temp:>7,.0f}
""")

if dosis_disponibles >= demanda_total_temp:
    excedente = dosis_disponibles - demanda_total_temp
    print(f"""
✓ RESULTADO: COBERTURA COMPLETA POSIBLE

  Excedente de dosis: +{excedente:,.0f} ({excedente/dosis_disponibles*100:.1f}%)

  RECOMENDACIÓN: Implementar estrategia completa (Escenarios 3 + 4)
""")
else:
    deficit = demanda_total_temp - dosis_disponibles
    if dosis_disponibles >= demanda_prioritaria_temp:
        excedente_prio = dosis_disponibles - demanda_prioritaria_temp
        print(f"""
⚠ RESULTADO: COBERTURA PARCIAL

  Déficit total: -{deficit:,.0f} dosis

  PERO: Se cubre 100% de población PRIORITARIA (prematuros)
        Excedente para adicionales: {excedente_prio:,.0f} dosis
        Cobertura adicional posible: {excedente_prio/demanda_adicional_temp*100:.1f}%

  RECOMENDACIÓN: Priorizar prematuros, cubrir a término según disponibilidad
""")
    else:
        cobertura_prio = dosis_disponibles / demanda_prioritaria_temp * 100
        print(f"""
✗ RESULTADO: DÉFICIT CRÍTICO

  Solo se cubre {cobertura_prio:.1f}% de población prioritaria

  RECOMENDACIÓN: Priorizar estrictamente:
    1. Muy prematuros (<32 sem)
    2. Prematuros (32-36 sem) según disponibilidad
""")

print("=" * 70)
print("ANÁLISIS COMPLETADO")
print("=" * 70)
```
